```python
import librosa
from librosa import stft
import numpy as np
import matplotlib.pyplot as plt
```

```python
sr = 200
t0 = 0
t1 = 3
x = np.linspace(t0, t1, sr*(t1-t0))
y = np.cos(x)
```

```python
win_samples = int(0.5*sr)
hop_length = win_samples # No overlap
D = stft(y, n_fft=win_samples, hop_length=hop_length, win_length=win_samples, window="blackman", center=False)
Sxx = np.abs(D)**2

n_frames = Sxx.shape[1]
edges = np.arange(n_frames + 1) * (hop_length / sr)
fig = librosa.display.specshow(Sxx, sr=sr, hop_length=hop_length, n_fft=win_samples, win_length = win_samples, x_axis='time', y_axis='hz')
plt.colorbar(fig)

plt.ylim(0,5)
plt.tight_layout()
plt.show()
```

```python
y, sr = librosa.load(librosa.example('brahms'))
import numpy as np
y = np.random.randn(int(sr * 2.0)) # 2 seconds of white noise
win_samples = int(0.5*sr)
hop_length = win_samples//2
D = stft(y, n_fft=win_samples, hop_length=hop_length, win_length=win_samples, window="boxcar", center=False)
Sxx = np.abs(D)**2

n_frames = Sxx.shape[1]
edges = np.arange(n_frames + 1) * (hop_length / sr)
fig = librosa.display.specshow(Sxx, sr=sr, hop_length=hop_length, n_fft=win_samples, win_length = win_samples, x_axis='time', y_axis='hz')
plt.colorbar(fig)

plt.ylim(0,2000)
plt.tight_layout()
plt.show()
```
